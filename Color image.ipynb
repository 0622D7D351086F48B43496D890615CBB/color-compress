```python
import  cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
image = cv2.imread("317.jpg.webp")
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
pixels = image.reshape((-1,3))
pixels = np.float32(pixels)
```

```python
print(pixels)
```

```python
k = 400
```

```python
when_stop = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER,100,0.2)
```

```python
Error,labels,centers = cv2.kmeans(pixels,k,None,when_stop,10,cv2.KMEANS_RANDOM_CENTERS)
```

```python
centers = np.uint8(centers)
```

```python
comperssed = centers[labels.flatten()]
```

```python
comperssed = comperssed.reshape(image.shape)
```

```python
cv2.imwrite("comperssed.jpg", cv2.cvtColor(comperssed, cv2.COLOR_RGB2BGR))
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(image)
plt.axis('off')
```

```python
plt.subplot(1,2,1)
plt.title("Comperssed (K=4)")
plt.imshow(comperssed)
plt.axis('off')

plt.show()
```
